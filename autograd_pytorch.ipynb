## Building a NN from scratch

```python
import torch

x=torch.tensor(6.7)
y=torch.tensor(0.0)

w=torch.tensor(1.0) #weights
b=torch.tensor(0) #bias
```

```python
def binary_cross_entropy_loss(prediction,target):
  epsilon=1e-8
  prediction=torch.clamp(prediction,epsilon,1-epsilon)
  return -target*torch.log(prediction)-(1-target)*torch.log(1-prediction)
```

```python
#forawrd pass
z=w*x+b
y_pred=torch.sigmoid(z)

loss=binary_cross_entropy_loss(y_pred,y)
```

```python
loss
```

```python
#backpropagation
dloss_dy_pred= (y_pred-y)/(y_pred*(1-y_pred))

dy_pred_dz= y_pred*(1-y_pred)

dz_dw=x
dz_db=1

dL_dw=dloss_dy_pred*dy_pred_dz*dz_dw
dL_db=dloss_dy_pred*dy_pred_dz*dz_db
```

```python
print("Manual Gradient of loss wrt weight: ",dL_dw)
print("Manual Gradient of loss wrt bias: ",dL_db)
```

### using autograd

```python
import torch

x=torch.tensor(6.7)
y=torch.tensor(0.0)

w=torch.tensor(1.0,requires_grad=True) #weights
b=torch.tensor(0.0,requires_grad=True) #bias
```

```python
z=w*x+b
y_pred=torch.sigmoid(z)

loss=binary_cross_entropy_loss(y_pred,y)
```

```python
loss.backward()
```

```python
w.grad,b.grad
```
